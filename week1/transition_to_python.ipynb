# Variable Scope

- Same name varibale in gloabl and local, will be treated two different variables.
- TO refer global variable inside the func, use `global` keyword


Nonlocal Keyword
- The nonlocal keyword is used to work with variables inside nested functions.

- The nonlocal keyword makes the variable belong to the outer function.



```python
x = 30
def myfunc():
  x = 300
  print(f"Inside func: {x}")

myfunc()

print(x) # Still 30 



def updateGlobalX():
  global x
  x = 300
  print(f"Inside func: {x}")

updateGlobalX()

print(x) # Still 30

def myfunc1():
  x = "Jane"
  def myfunc2():
    nonlocal x
    x = "hello"
  myfunc2()
  return x

print(myfunc1())
```

The LEGB Rule

- Python follows the LEGB rule when looking up variable names, and searches for them in this order:

- Local - Inside the current function
- Enclosing - Inside enclosing functions (from inner to outer)
- Global - At the top level of the module
- Built-in - In Python's built-in namespace

```python
x = "global"

def outer():
  x = "enclosing"
  def inner():
    x = "local"
    print("Inner:", x)
  inner()
  print("Outer:", x)

outer()
print("Global:", x)
```

# Comprehensions

Comprehensions in Python provide a concise, readable, and highly efficient way to create new collections (lists, dictionaries, sets, or generators) from existing iterables. They serve as a "Pythonic" shortcut that replaces traditional for loops and .append() blocks into a single line of code.

```python
# List Comprehension
# Traditional way
squares = []
for x in range(5):
    squares.append(x**2)

# Comprehension way
squares = [x**2 for x in range(5)]
# Output: [0, 1, 4, 9, 16]


# Create a dictionary of number: square mapping
square_dict = {x: x**2 for x in range(5)}
# Output: {0: 0, 1: 1, 2: 4, 3: 9, 4: 16}


# Extract unique characters from a string
unique_chars = {char for char in "banana"}
# Output: {'b', 'a', 'n'}


# Creates a generator object
# Creates a memory-efficient, lazy-loading iterator rather than building the entire collection in memory immediately
lazy_squares = (x**2 for x in range(1_000_000_000))

print(f"Sum using generator objects: {sum(lazy_squares)}")

# The example above still requires a lot of work, but it performs the operations lazily. 
# Because of lazy evaluation, your code only calculates values when they’re explicitly requested. 
# After the generator yields a value, it can add that value to the running sum, 
# then discard that value and generate the next value. When the sum() function requests the next value, 
# the cycle starts over. This process keeps the memory footprint small.

# sum([x for x in range(1_000_000_000)]) # Heavy in memory 


# The map() function also operates lazily, meaning memory won’t be an issue if you choose to use it in this case:

sum(map(lambda x: x**2, range(1_000_000_000)))
```


# What is a Closure?

- A closure is an inner function that remembers and has access to variables from its outer (enclosing) function's scope, even after the outer function has finished executing.

- To create a closure, you must meet three criteria:
1. Have a nested function (a function inside a function).
2. The nested function must refer to a variable defined in the enclosing scope.
3. The enclosing function must return the nested function.

```python
def power_factory(exponent):
    # 'exponent' is a local variable to power_factory
    def calculate_power(base):
        # The inner function remembers 'exponent' even after power_factory finishes
        return base ** exponent
    return calculate_power

# Create specialized functions
square = power_factory(2)
cube = power_factory(3)

print(square(5))  # Output: 25
print(cube(5))    # Output: 125


# Why it works: When power_factory(2) executes, its scope normally destroys itself. 
# However, because calculate_power references exponent, Python bundles that environment into the function object. 
# You can inspect this hidden state using square.__closure__.
```

# Python Decorators

- Decorators let you add extra behavior to a function, without changing the function's code.

- A decorator is a function that takes another function as input and returns a new function.



## What is a Decorator?

- A decorator is a design pattern that allows you to extend or modify the behavior of a function or method without permanently changing its source code.
- Under the hood, a decorator is a closure where the enclosed variable being "remembered" is a function object.

- Instead of manually passing a function into a closure, Python provides the @ syntax as "syntactic sugar" to make wrapping functions cleaner:


```python
def toupper(func):
    def myinner():
        return func().upper()
    return myinner

@toupper
def decorated_func():
    print(3)
    return "Umesh Kumar"


print(decorated_func())

########################################
# Real world example of decorator
import time
from functools import lru_cache


@lru_cache              # Use lru_cache as decorator
def fibonacci(n):

    if n < 2:
        return n
    return fibonacci(n-1) + fibonacci(n-2)
start_time = time.perf_counter()
for i in range(35):
    print(f"{fibonacci(i)}", end=" ")

end_time = time.perf_counter()
print(f"Execution time: {end_time - start_time}")


```

```python
import functools

def my_logger(func):
    @functools.wraps(func) # Preserves the original function's name and docstring
    def wrapper(*args, **kwargs):
        print(f"--> Running '{func.__name__}' with arguments: {args}")
        result = func(*args, **kwargs)
        print(f"--> '{func.__name__}' finished execution.")
        return result
    return wrapper

@my_logger
def add_numbers(a, b):
    return a + b

# Calling the decorated function
print(add_numbers(5, 10))
print(add_numbers.__closure__)
```

```python
import copy

# A nested list: an outer list containing an inner mutable list
original = [[1, 2, 3], "static_string"]

# Create copies
shallow_copied = copy.copy(original)
deep_copied = copy.deepcopy(original)

# --- Test 1: Modifying the outer container ---
original.append("new_item")
print("After appending to outer list:")
print(f"Original: {original}")        # [[1, 2, 3], 'static_string', 'new_item']
print(f"Shallow:  {shallow_copied}")   # [[1, 2, 3], 'static_string'] -> Unaffected!
print(f"Deep:     {deep_copied}")      # [[1, 2, 3], 'static_string'] -> Unaffected!

print("\n" + "="*40 + "\n")

# --- Test 2: Modifying a nested mutable object ---
original[0].append(99)
print("After appending to nested list:")
print(f"Original: {original}")        # [[1, 2, 3, 99], 'static_string', 'new_item']
print(f"Shallow:  {shallow_copied}")   # [[1, 2, 3, 99], 'static_string'] -> MODIFIED!
print(f"Deep:     {deep_copied}")      # [[1, 2, 3], 'static_string'] -> SAFE!
```

# Generators
- Generators are functions that can pause and resume their execution.

- When a generator function is called, it returns a generator object, which is an iterator.

- The code inside the function is not executed yet, it is only compiled. The function only executes when you iterate over the generator.



```python
def fibonacci():
  a, b = 0, 1
  while True:
    yield a
    a, b = b, a + b

# Get first 100 Fibonacci numbers
gen = fibonacci()
for _ in range(100):
  print(next(gen))
```
